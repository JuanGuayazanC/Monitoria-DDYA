### Conjunto disyunto

Como recordatorio, tenemos a disposicion la clase `DisjointSet`, y las operaciones:
- `make_set(value: Any) -> DisjointSet`: Crea un nuevo conjunto disyunto
- `find_set(x: DisjointSet) -> DisjointSet`: Encuentra el representante del conjunto $x$
- `is_same_set(x: DisjointSet, y: DisjointSet) -> bool`: Verifica si el el elemento $x$ y $y$, pertenecen al mismo conjunto.
- `union(x: DisjointSet, y: DisjointSet)`: Une dos conjuntos $x$ y $y$ en uno solo

```python
# @title Ejecute esta celda para definir `DisjointSet`

from typing import Any

class DisjointSet:
    def __init__(self, value: Any) -> None:
        self.value: Any = value
        self.parent: DisjointSet = self
        self.rank: int = 0

def make_set(value: Any) -> DisjointSet:
    return DisjointSet(value)

def find_set(x: DisjointSet) -> DisjointSet:
    if x != x.parent:
        x.parent = find_set(x.parent)

    return x.parent

def is_same_set(x: DisjointSet, y: DisjointSet) -> bool:
    return find_set(x) == find_set(y)

def link(x: DisjointSet, y: DisjointSet):
    if x.rank > y.rank:
        y.parent = x
    else:
        x.parent = y
        if x.rank == y.rank:
            y.rank += 1

def union(x: DisjointSet, y: DisjointSet):
    link(find_set(x), find_set(y))
```

### Cola de prioridad _minimal_

Tenemos a disposicion la clase `MinPriorityQueue` que contiene las operaciones:
- `insert(value: Any)`: Inserta un nuevo valor a nuestra cola
- `extract_min() -> Any`: Extrae el elemento con el valor mas pequeño
- `get_min() -> Any`: Retorna el elemento mas pequeño, sin extraerlo de la cola

```python
# @title Ejecute esta celda para definir `MinPriorityQueue`

from typing import Any
import heapq

class MinPriorityQueue:
  def __init__(self, content: list[Any] = []):
    self.heap: list[Any] = content.copy()
    heapq.heapify(self.heap)

  def insert(self, value: Any):
    heapq.heappush(self.heap, value)

  def extract_min(self) -> Any:
    assert len(self.heap) > 0, "la cola esta vacia"
    v = heapq.heappop(self.heap)
    return v

  def get_min(self) -> Any:
    assert len(self.heap) > 0, "la cola esta vacia"
    return self.heap[0]
```
