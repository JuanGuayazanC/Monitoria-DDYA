## Graficas

Debido a que los cuadernos de Jupyter estan pensados para tareas mucho mas complejas, tambien podemos hacer cosas como graficas con codigo, incluso hacerlas interactivas en ciertas ocaciones!

Observe que ocurre al ejecutar las siguientes dos (2) casillas.

````python
def dibujar(x, y, dimensiones=(7, 5)):
  """
  Dibuja una serie de puntos en un plano cartesiano

  x: lista de numeros
  y: lista de numeros
  """
  assert len(x) == len(y), "la cantidad de numeros en ambas debe ser el mismo"

  import matplotlib.pyplot as plt
  plt.figure(figsize=dimensiones)
  plt.plot(x, y)
  plt.show()
````

````python
x_0 = 0.0 # punto inicial
x_f = 2.0 # punto final
dx = 0.01 # tamaño de intervalo

N = int((x_f - x_0) / dx) # numero de puntos que debemos graficar
x = [ x_0 + dx * i for i in range(N+1) ]
y = [ v * v for v in x ]

# Dibujamos nuestros puntos que estan en el eje 'x' y 'y'
dibujar(x, y)
````

Como puede observar, se genero una grafica de una curva de la funcion $x^2$ en el rango $[0, 2]$ en intervalos de $0.01$.

> En esta seccion daremos la especificacion del problema a resolver, en este no queremos dejar detalles del enunciado sino solo detalles relevantes para la solucion del problema, en ocasiones podemos usar vocabulario formal para esta seccion. **Problema:** Saltos altos y bajos.

**Entrada:** Una secuencia de $N$ numeros enteros positivos $\langle a_1, \ldots, a_{N} \rangle$.

**Salida:** Dos numeros enteros que describen que para cualquier par de numeros consecutivos $a_i$ y $a_{i+1}$, muestran la cantidad de veces en las que se tiene $(a_i < a_{i+1})$ y $(a_i > a_{i+1})$ respectivamente.

> Notese que en esta especificacion no hicimos mencion del termino $T$ descrito en el enunciado del problema, esto es por que solo nos interesa describir la solucion del problema planteado, no queremos dejar detalles tecnicos de la entrada ni de la salida de datos en esta parte.

## Diseño

> En esta seccion escribiremos todos los detalles de diseño que utilizamos para la solucion de nuestro problema, puede incluir imagenes, formulas, entre otros segun se considere necesario.

El problema nos pide hallar la cantidad de veces en las que Mario debe tomar saltos hacia arriba o hacia abajo dadas las alturas de las diferentes paredes que este debe pasar para finalizar su mision.

Podemos contemplar tres (casos) distintos a tener en cuenta:
1. La pared a la que va a saltar esta a la misma altura que la actual ($a_i = a_{i+1}$)
2. La pared a la que va a saltar esta a *mayor* altura que la actual ($a_i < a_{i+1}$)
3. La pared a la que va a saltar esta a *menor* altura que la actual ($a_i > a_{i+1}$)

Solo nos interesa tener un conteo de los casos 2 y 3.

image.png

Para resolver este problema, debemos de ir por cada par de elementos en la secuencia $a_i$ y $a_{i+1}$ para algun $i$ en orden secuencial, es decir $i = 1, 2, 3, \ldots, (N-1)$ y mantener un conteo de los saltos altos ($\text{high}$) y bajos ($\text{low}$), ambos con valores iniciales de $0$.

En cada iteracion $i$, debemos evaluar cual de los 3 casos es para saber que accion tomar:

1. En este caso decidimos avanzar sin modificar $\text{high}$ ni $\text{low}$ dado que no cuenta como un salto para arriba ni para abajo.
2. Decidimos avanzar modificando $\text{high}$ por medio de aumentar su valor $+1$
3. Decidimos avanzar modificando $\text{low}$ por medio de aumentar su valor $+1$

Finalizamos la iteracion, y ahora $i$ toma el valor de $i+1$, tomamos esta accion hasta que lleguemos a $i = N-1$.

## Analisis de complejidad

> En esta seccion haremos un analisis de los tiempos de ejecucion con tecnicas que veremos en mas detalle mas adelante en el curso, por ahora interpretelo como el analisis de tiempo que toma nuestro programa en correr.

La cantidad de operaciones que debemos hacer depende enteramente en las iteraciones que debemos realizar sobre nuestra secuencia de entrada.

En cada una de las iteraciones realizamos evaluaciones y operaciones que toman tiempo constante.

Por tanto, la cantidad de tiempo que toma nuestra solucion en ejecutarse es lineal, es decir, para nuestra entrada de tamaño $N$, nuestra solucion tomaria una forma de $T(N) = cN$, donde $c$ es una constante positiva.

> La justificacion del tiempo de ejecucion en este ejemplo es intencionalmente vaga ya que esto hace parte de lo que trabajaremos en mas profundidad en las siguientes clases, donde haremos una explicacion mas profunda de los tiempos de ejecucion y utilizaremos otra notacion para denotarla.

## Codigo

> En esta seccion especificaremos el codigo con el que podemos solucionar nuestro problema, usando el diseño planteado en la seccion anterior para guiarnos.

````python
# Funcion que contiene nuestra solucion
def solve(a: list[int]) -> tuple[int, int]:
  n = len(a)
  high, low = 0, 0
  for i in range(n-1):
    if a[i] == a[i+1]:
      # El primer caso, no hacemos nada
      pass
    elif a[i] > a[i+1]:
      low += 1
    else: # solo puede ser a[i] < a[i+1]
      high += 1
  return high, low
````

## Pruebas de enunciado

> En esta subseccion haremos pruebas de los casos de prueba tal como salen en los ejemplos del enunciado. Tambien puede optar por agregar casos adicionales de prueba para evaluarlos y verificar manualmente que son correctos.

````python
# Definimos una funcion de prueba facil de nuestros casos
def test(input: list[int], expected_output: tuple[int, int]):
  """
  input: entrada de datos
  expected_output: salida esperada de datos
  """
  # Notese que hacemos la prueba directamente sobre 'solve'
  got_output = solve(input)
  assert got_output == expected_output, f"se obtuvo {got_output}, pero se esperaba {expected_output}"
  print(f"Paso caso de prueba {input} -> {expected_output}")
````

## Pruebas de rendimiento

> En esta seccion intentaremos mostrar graficamente los tiempos de ejecucion de nuestro algoritmo y veremos si en efecto se comporta como planteamos en la seccion de `Analisis de complejidad`

````python
# Calcula el tiempo de ejecucion con base a diferentes tamaños de la entrada
# y nos da una lista de resultados, cada resultado contiene el tamaño de la entrada
# y el tiempo de ejecucion
def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
  print("Generando datos ...", end='')
  from random import randint # para generar numeros aleatorios
  import time
  results: list[tuple[int, float]] = []
  for size in range(1, n, step):
    input = [randint(0, 10) for _ in range(size)]
    ti = time.time()
    output = solve(input)
    tf = time.time()
    delta = tf - ti
    print(".", end="")
    results.append((size, delta)) # Guardamos los tamaños de la entrada y el tiempo que tomo en ejecutarse

  print("Finalizado")
  return results
````

````python
tiempos = bench(10_000_000, step = 250_000)

# Los convertimos a coordenadas para dibujar una grafica
x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]

dibujar(x, y)
````

Como se puede observar de la grafica generada, el comportamiento de nuestra solucion es lineal, con ciertos casos atipicos causado por factores externos del entorno de ejecucion.
