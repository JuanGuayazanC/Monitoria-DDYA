## Arbol binario

Un árbol binario es una estructura de datos jerárquica en la que cada nodo tendrá **dos** hijos, referidos como árbol izquierdo (`left`) y arbol derecho (`right`). Cada nodo del arbol contiene una valor asociado (`key`) que puede contener cualquier tipo de dato internamente.

Cualquiera de los árboles pueden ser nulos, referidos en el libro guia como `NIL` y en Python como `None`.

Para poder trabajar con esta estructura de datos, podemos definirla de la siguiente manera en Python, como lo hicimos con nuestro autoestudio anterior de estructuras lineales.

````python
class Tree:
  def __init__(self, key):
    self.key = key # valor que guardamos
    self.left: Tree | None = None
    self.right: Tree | None = None
    self.parent: Tree | None = None

  def __repr__(self) -> str:
    return f"[{self.key}]"
````

Dentro de la definicion que hicimos de `Tree`, notese que para el árbol izquierdo y derecho tenemos definido el tipo de datos guardados como `Tree | None`, esto significa que el valor que guarda nuestra variable es un nodo de árbol o bien es la ausencia de uno (`None`).

Y finalmente adicionamos el `parent` que apunta al ancestro del nodo de árbol. La raiz del árbol es el unico nodo que no tiene ancestro.

Para nosotros instanciar un árbol podemos hacerlo de la siguiente manera:

````python
arbol: Tree | None = None # Inicialmente vacio o nulo
arbol = Tree(5) # arbol con valor guardado de `key = 5`
arbol_izquierdo = Tree(6) # arbol que vamos a insertar como árbol izquierdo
arbol_derecho = Tree(4) # arbol que vamos a insertar como árbol derecho

arbol.left = arbol_izquierdo
arbol.right = arbol_derecho
````

Ahora, ¿Como podriamos definir la forma de imprimir el arbol en texto? Para esto podemos aprovechar los diferentes tipos de recorrido que existen para arboles:
 - **Inorden**: arbol izquierdo, raiz, arbol derecho
 - **Preorden**: raiz, arbol izquierdo, arbol derecho
 - **Posorden**: arbol izquierdo, arbol derecho, raiz

Podemos definir un metodo para realizar la exploracion en _inorden_ de la siguiente manera:

````python
def inorder_walk(x: Tree | None) -> list:
  if x is None:
    return []

  walk = []
  walk += inorder_walk(x.left)
  walk.append(x.key)
  walk += inorder_walk(x.right)

  return walk
````

## Arbol binario de busqueda (BST)

Un árbol binario de busqueda (BST por sus siglas en ingles), es un árbol que adicional a ser binario cumple la **propiedad de árbol binario de busqueda** definida como:

> Sea $y$ un nodo en un árbol binario de busqueda y sean $x$ y $z$ los subarboles izquierdo y derecho respectivamente de $y$, se cumple que $x.key \leq y.key \leq z.key$.

Ahora tambien podemos proceder a definir metodos sobre nuestra estructura de datos que nos permitiran realizar consultas y modificaciones manteniendo la propiedad de árbol binario de busqueda.

Empecemos con una operacion de consulta: **busqueda**. Para esto definamos el metodo `search_tree` que recibira como entrada un arbol `x` en donde realizaremos la busqueda y el valor del nodo `k` del arbol que buscaremos. Este metodo debera devolver el nodo del árbol que cumple `x.key == k`, en caso de no encontrarse debera devolver `None`.

````python
def search_tree(x: Tree | None, k) -> Tree | None: # O(h)
  if x is None:
    return None

  if x.key == k:
    # encontramos el nodo
    return x

  if k < x.key:
    return search_tree(x.left, k)
  else:
    return search_tree(x.right, k)
````

Ahora definamos un arbol sencillo para probar nuestro metodo:

````python
arbol = Tree(3)
arbol.left = Tree(1)
arbol.left.left = Tree(0)
arbol.left.right = Tree(2)
arbol.right = Tree(4)
arbol.right.left = Tree(3)
arbol.right.right = Tree(5)

# Ahora dibujemoslo
draw(arbol)
````

````python
# Ahora busquemos el subarbol con valor de 5
z = search_tree(arbol, 5)
print(z.key)
draw(z)
````

````python
# Ahora busquemos el subarbol con valor de 4
z = search_tree(arbol, 4)
print(z.key)
draw(z)
````

````python
# Y finalmente busquemos algo que no existe en el arbol
z = search_tree(arbol, 6)
print(z)
````

## Insercion de elementos

Ahora que tenemos formas de realizar la busqueda sobre un árbol binario de busqueda, nos queda tambien realizar la insercion de un elemento respetando la propiedad de árbol binario de busqueda.

Para esto la idea es muy sencilla:
1. Debemos encontrar el punto de insercion de nuestro elemento: el punto en donde no podemos descender mas en nuestro árbol.
2. Insertar el nuevo nodo en el subárbol correspondiente del nodo que encontramos.

Definamos dos metodos para realizar la insercion:
1. `find_insertion_point`
2. `insert_node`

El metodo principal para la insercion es `tree_insert`.

````python
def find_insertion_point(t: Tree, z: Tree) -> Tree:
  assert t is not None
  assert z is not None

  subtree: Tree | None = None
  if z.key < t.key:
    subtree = t.left
  else:
    subtree = t.right

  if subtree is None:
    # No podemos descender mas por que el arbol donde debemos insertar es vacio,
    # entonces 't' es el nodo donde debemos insertar
    return t

  return find_insertion_point(subtree, z)
````

Asumamos que tenemos el siguiente arbol

````python
t = Tree(6)
t.left = Tree(3)
t.left.left = Tree(2)
t.left.right = Tree(4)
t.left.right.left = Tree(3)
t.right = Tree(7)

draw(t)
````

Ahora asumamos que vamos a insertar un elemento con valor de `5`, primero debemos determinar el punto de insercion.

````python
new_node = Tree(5)
insertion_point = find_insertion_point(t, new_node)
print(insertion_point)
draw(insertion_point)
````

Ahora que encontramos el punto de insercion (que es el nodo `4`), debemos determinar si debemos agregar el nuevo nodo como árbol izquierdo o derecho. Esta decision la tomamos en el metodo `insert_node`.

````python
def insert_node(insertion_point: Tree, z: Tree):
  z.parent = insertion_point # el nuevo nodo tiene como ancestro el punto de insercion

  # Determinamos si debemos insertar el nuevo nodo en la izquierda o la derecha del punto de insercion
  if z.key < insertion_point.key:
    insertion_point.left = z
  else:
    insertion_point.right = z
````

````python
insert_node(insertion_point, new_node)
# Visualizamos desde el punto de insercion
draw(insertion_point)
````

````python
# Ahora visualicemos desde la raiz
draw(t)
````

````python
def tree_insert(t: Tree | None, z: Tree) -> Tree:
  if t is None:
    # Caso especial: el arbol original es vacio, por lo que el nuevo arbol es 'z'
    return z

  insertion_point = find_insertion_point(t, z)
  insert_node(insertion_point, z)

  return t
````

Ahora pongamos a prueba nuestro codigo.

Ahora definimos el metodo `tree_insert` el cual se encarga de llamar a los dos metodos anteriormente definidos. Este tiene como entrada el árbol donde vamos a insertar `t`, y el nuevo nodo que vamos a insertar `z`, este devuelve la raiz del árbol, este nos sirve para simplificar el caso especial en donde el árbol es vacio (`None`).

````python
t = None # Iniciamos con un arbol vacio

for value in [12, 5, 18, 2, 9, 15, 19, 13, 17]:
  t = tree_insert(t, Tree(value))
  print(inorder_walk(t))

draw(t)
````

Esta definicion que hemos hecho es _poderosa_, ya que nuestro metodo de insercion habla en terminos de agregar un árbol completo, lo que significa que tambien nos permite agregar un árbol a otro árbol tambien y seguiria siendo un árbol binario de busqueda.

````python
a = None # Primer arbol
for value in [7, 4, 3, 6, 2]:
  a = tree_insert(a, Tree(value))

b = None # Segundo arbol
for value in [11, 9, 18, 14, 19, 12, 17, 22, 20]:
  b = tree_insert(b, Tree(value))

print("a =", inorder_walk(a))
print("b =", inorder_walk(b))

a = tree_insert(a, b)
print("a+b =", inorder_walk(a))

draw(a)
````

## ¿Los árboles son iguales?

Consideremos el caso en la que queramos saber si un arbol es igual a otro. En este momento, por lo que hemos definido de nuestra clase de Python `Tree`, la pregunta de `Tree(a) == Tree(b)` solo seria afirmativa si se tratase de exactamente el mismo objeto, es decir, si el identificador interno del objeto es igual para ambos objetos.

````python
a = Tree(1)
b = Tree(1)

print("a == a", a == a)
print("a == b", a == b)
print("b == b", b == b)
print(f"id(a) = {id(a)}, id(b) = {id(b)}")
````

Si quisieramos de verdad comprobar si un árbol es igual a otro, una forma que podriamos hacerlo es solo considerar que son iguales si la llave es la misma y tambien si los subárboles son iguales. Podemos definirlo como:

````python
def trees_are_equal(a: Tree | None, b: Tree | None) -> bool:
  if a is None and b is None:
    # si ambos arboles son vacios
    return True

  if a is None or b is None:
    # solo uno de los arboles es vacio
    return False

  if a.key != b.key:
    # Si las llaves no coinciden, entonces no son iguales
    return False

  # Optimizacion para no verificar todos los subarboles: ¿son los identificadores de los arboles iguales?
  if id(a) == id(b):
    return True

  # Solo pueden ser iguales si los subarboles son iguales
  left_is_equal = trees_are_equal(a.left, b.left)
  right_is_equal = trees_are_equal(a.right, b.right)

  return left_is_equal and right_is_equal
````

````python
# Hagamos una prueba sencilla
a = Tree(3)
a.left = Tree(2)
a.left.left = Tree(1)
a.left.right = Tree(2)
a.right = Tree(4)

b = Tree(3)
b.left = Tree(2)
b.left.left = Tree(1)
b.left.right = Tree(2)
b.right = Tree(4)

c = Tree(2)
c.left = Tree(2)
c.left.left = Tree(1)
c.left.right = Tree(2)
c.right = Tree(3)
c.right.right = Tree(4)

draw(a)
````

````python
draw(b)
````

````python
draw(c)
````

````python
# Comprobemos la igualdad de cada uno

print("a == a", trees_are_equal(a, a))
print("a == b", trees_are_equal(a, b))
print("a == c", trees_are_equal(a, c))
print("b == c", trees_are_equal(b, c))
````

Ahora bien, si recordamos el autoestudio de _estructuras lineales_, podemos definir el metodo de igualdad entre dos objetos de la misma clase por medio de definir el metodo `__eq__`. Ya que ya implementamos el metodo de igualdad con `trees_are_equal`, podemos sencillamente reutilizarlo para nuestros fines y definir el operador de igualdad `==` de la siguiente manera.

**Nota**: podemos hacer esto ya que Python espera que el operador de igualdad tenga la firma `__eq__(a, b) -> bool`.

````python
Tree.__eq__ = trees_are_equal
````

````python
# Ahora si podemos hacer uso del operador '=='

print("a == a", a == a)
print("a == b", a == b)
print("a == c", a == c)
print("b == c", b == c)
````

## Minimo y maximo

Dentro de un árbol binario de busqueda, podemos encontrar facilmente el minimo y el maximo.

Basta con explorar los subárboles izquierdos hasta encontrar un subárbol que no contenga árbol izquierdo.

Podemos definir el metodo `minimum` de la siguiente manera:

````python
def minimum(x: Tree | None) -> Tree | None:
  if x is None:
    return None

  while x.left is not None:
    x = x.left
  return x
````

## Sucesor y predecesor

Considerando la exploracion en _inorden_ de un arbol binario de busqueda, el sucesor de un elemento dentro del arbol es el elemento siguiente al elemento en el _inorden_.

Por ejemplo, dado el inorden de un BST:
```
[1, 3, 4, 6, 7]
```
El sucesor del elemento con el valor de `4` es el nodo con el valor de `6`. El sucesor de `6` es el `7`. El sucesor de `1` es `3`.

Para determinar el sucesor de un elemento `x` en el arbol existen dos casos que debemos tener en cuenta.

1. Si existe un arbol derecho `x.right`, el sucesor es el nodo con el valor minimo del arbol derecho, es decir `minimum(x.right)`
2. Si no existe arbol derecho, debemos explorar dentro de nuestros ancestros y determinar el primer ancestro en el cual nuestro nodo pertenece al subarbol izquierdo.

````python
def successor(x: Tree) -> Tree | None:
  if x.right is not None:
    # tenemos un arbol derecho, entonces ...
    return minimum(x.right)

  # de lo contrario, buscamos en nuestros ancestros
  y = x.parent # 'y' es nuestro ancestro
  while y is not None and x == y.right:
    # mientras tengamos ancestro y seamos parte del subarbol izquierdo
    x = y
    y = y.parent
  return y
````
