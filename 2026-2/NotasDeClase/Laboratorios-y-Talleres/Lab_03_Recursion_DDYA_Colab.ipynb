# Recordatorio: una función recurrente se define por casos

Una definición recursiva puede escribirse con la misma notación de las funciones a trozos:

$$
f(n)=
\begin{cases}
\text{valor directo}, & \text{si } n \text{ cumple el caso base},\\
\text{combinar}\bigl(n,f(r(n))\bigr), & \text{en otro caso}.
\end{cases}
$$

La condición del segundo caso debe acercarnos al caso base. Antes de escribir Python, completa esta estructura para cada problema.
