# Ejemplo del formato · Insertion Sort

Este ejemplo muestra cómo implementar una función, comprobarla y analizarla. Las soluciones de los cuatro problemas del taller no están incluidas.

## Implementación del ejemplo


````python
def findSortedPlace(data, el):
    """Inserta el en la posición correspondiente de una secuencia ordenada."""
    index = 0
    while index < len(data) and data[index] <= el:
        index += 1
    return data[:index] + [el] + data[index:]


def insertionSort(data):
    """Retorna los elementos de data ordenados incrementalmente."""
    initial_index = 1
    sorted_data = data[:initial_index]
    for index in range(initial_index, len(data)):
        sorted_data = findSortedPlace(sorted_data, data[index])
    return sorted_data
````

## Comprobación del ejemplo


````python
print(insertionSort([]) == [])
print(insertionSort([8]) == [8])
print(insertionSort([1, 2, 3, 4]) == [1, 2, 3, 4])
print(insertionSort([4, 3, 2, 1]) == [1, 2, 3, 4])
print(insertionSort([3, 1, 3, 2]) == [1, 2, 3, 3])
````

## Análisis asintótico del ejemplo

El análisis se escribe en una celda de Python, directamente al lado de cada instrucción.


````python
# Pre: data está ordenada y el no pertenece a data
# Post: el resultado contiene el, conserva los elementos de data y está ordenado
def findSortedPlace(data, el):                        # Costo peor | Veces peor | Costo mejor | Veces mejor
    index = 0                                         # c0          | 1           | c0          | 1
    while index < len(data) and data[index] <= el:    # c1          | n + 1       | c1          | 1
        index += 1                                    # c2          | n           | c2          | 0
    return data[:index] + [el] + data[index:]         # c3          | 1           | c3          | 1

# T_peor(n) = c0 + c1(n+1) + c2n + c3 = O(n)
# T_mejor(n) = c0 + c1 + c3 = Ω(1)

# Pre: data contiene elementos comparables
# Post: el resultado contiene los elementos de data y está ordenado
def insertionSort(data):                              # Costo peor | Veces peor | Costo mejor | Veces mejor
    initial_index = 1                                 # c0          | 1           | c0          | 1
    sorted_data = data[:initial_index]                # c1          | 1           | c1          | 1
    for index in range(initial_index, len(data)):     # c2          | n           | c2          | n
        sorted_data = findSortedPlace(                # n           | n - 1       | 1           | n - 1
            sorted_data, data[index])
    return sorted_data                                # c3          | 1           | c3          | 1

# T_peor(n) = O(n²)
# T_mejor(n) = Ω(n)
````

> Para este taller se usa el modelo de costos mostrado en clase. En una medición específica de Python, los cortes de listas también dependen de la cantidad de elementos copiados.
