## Concepto 1: asignación


La asignación utiliza `=` para guardar un valor en una variable. Si una
variable aparece nuevamente a la izquierda, su valor anterior se reemplaza.
Las expresiones de la derecha se evalúan antes de realizar la asignación.

````python
precio = 2500
cantidad = 3
subtotal = precio * cantidad
cantidad = cantidad + 1

print("Subtotal inicial:", subtotal)
print("Nueva cantidad:", cantidad)
````

**Salida esperada:**

```text
Subtotal inicial: 7500
Nueva cantidad: 4
```

## Concepto 2: condicionales


Los condicionales permiten seleccionar instrucciones según una expresión
booleana. En una estructura `if`/`elif`/`else` solamente se ejecuta la primera
rama cuya condición sea verdadera.

````python
temperatura = 18

if temperatura < 10:
    mensaje = "Hace frío"
elif temperatura < 25:
    mensaje = "Temperatura moderada"
else:
    mensaje = "Hace calor"

print(mensaje)
````

**Salida esperada:**

```text
Temperatura moderada
```

## Concepto 3: ciclos


Un ciclo repite instrucciones. `range(1, 5)` produce los valores 1, 2, 3 y 4:
el límite superior no se incluye. Un acumulador debe inicializarse antes del
ciclo y actualizarse dentro de cada iteración.

````python
total = 0

for numero in range(1, 5):
    total = total + numero
    print("Número:", numero, "- Total:", total)

print("Suma final:", total)
````

**Salida esperada:**

```text
Número: 1 - Total: 1
Número: 2 - Total: 3
Número: 3 - Total: 6
Número: 4 - Total: 10
Suma final: 10
```

## Concepto 4: listas o arreglos


Una lista almacena varios valores en orden. Los índices comienzan en cero.
Puede recorrerse con `for`, consultarse con `lista[indice]` y ampliarse con
`append`.

````python
notas = [3.5, 4.2, 2.8]
notas.append(4.5)

print("Primera nota:", notas[0])
print("Cantidad:", len(notas))

for nota in notas:
    print("Nota:", nota)
````

**Salida esperada:**

```text
Primera nota: 3.5
Cantidad: 4
Nota: 3.5
Nota: 4.2
Nota: 2.8
Nota: 4.5
```

## Concepto 5: matrices


Una matriz puede representarse como una lista de listas. El primer índice
selecciona la fila y el segundo la columna. Para recorrer todos sus elementos
se utilizan normalmente dos ciclos anidados.

````python
matriz = [
    [2, 4, 6],
    [1, 3, 5],
]

print("Elemento de fila 1, columna 2:", matriz[0][1])

for fila in matriz:
    suma_fila = 0
    for valor in fila:
        suma_fila = suma_fila + valor
    print("Suma de la fila:", suma_fila)
````

**Salida esperada:**

```text
Elemento de fila 1, columna 2: 4
Suma de la fila: 12
Suma de la fila: 9
```

## Concepto 6: tuplas


Una tupla agrupa valores ordenados y utiliza normalmente paréntesis. Permite
índices y desempaquetado, pero es inmutable: después de crearla no se pueden
reemplazar, agregar o eliminar sus elementos.

````python
coordenada = (4, 7)
x, y = coordenada

print("Primer elemento:", coordenada[0])
print("x:", x)
print("y:", y)
print("Cantidad de elementos:", len(coordenada))
````

**Salida esperada:**

```text
Primer elemento: 4
x: 4
y: 7
Cantidad de elementos: 2
```

## Concepto 7: asignación múltiple o vectorial


Python permite asignar varios valores en una sola instrucción. La cantidad de
variables debe coincidir con la cantidad de valores. Esta sintaxis también
permite intercambiar variables sin utilizar una variable temporal.

````python
nombre, edad, activo = "Ana", 19, True
a, b = 5, 9
a, b = b, a

print(nombre, edad, activo)
print("a:", a)
print("b:", b)
````

**Salida esperada:**

```text
Ana 19 True
a: 9
b: 5
```

## Concepto 8: recorrido directo y recorrido indexado


Cuando solamente se necesita cada elemento, se recomienda recorrer la lista de
forma directa. Cuando también se necesita la posición, puede recorrerse
`range(len(lista))` y usar el índice para acceder al elemento.

````python
frutas = ["pera", "mango", "uva"]

print("Recorrido directo:")
for fruta in frutas:
    print(fruta)

print("Recorrido indexado:")
for indice in range(len(frutas)):
    print(indice, frutas[indice])
````

**Salida esperada:**

```text
Recorrido directo:
pera
mango
uva
Recorrido indexado:
0 pera
1 mango
2 uva
```

## Concepto 9: funciones y firmas


Una función agrupa instrucciones reutilizables. Su firma incluye el nombre,
los parámetros y, cuando corresponde, valores predeterminados. En la firma
`calcular_area(base, altura=1)`, `base` es obligatorio y `altura` es opcional.
`return` entrega el resultado a quien llamó la función.

````python
def calcular_area(base, altura=1):
    return base * altura


area_rectangulo = calcular_area(5, 3)
area_con_altura_predeterminada = calcular_area(base=4)

print(area_rectangulo)
print(area_con_altura_predeterminada)
````

**Salida esperada:**

```text
15
4
```

## Concepto 10: alcance de las variables


El alcance determina dónde puede utilizarse una variable. Una variable creada
dentro de una función es local. Si tiene el mismo nombre que una variable
externa, la variable local la oculta únicamente mientras se ejecuta la función.

````python
mensaje = "valor global"


def mostrar_mensaje():
    mensaje = "valor local"
    print("Dentro:", mensaje)


mostrar_mensaje()
print("Fuera:", mensaje)
````

**Salida esperada:**

```text
Dentro: valor local
Fuera: valor global
```
