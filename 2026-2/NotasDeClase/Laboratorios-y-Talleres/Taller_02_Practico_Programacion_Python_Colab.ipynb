## Ejemplo: cómo utilizar la entrada y la salida

El siguiente ejemplo solicita una cantidad con `input`. Como `input` produce
texto, se utiliza `int` para convertirlo en un número entero. Luego se calcula
el total y se presenta con `print`.

````python
cantidad = int(input("Cantidad de cuadernos: "))
precio_unitario = 3500
total = cantidad * precio_unitario

print("Cantidad comprada:", cantidad)
print("Total a pagar:", total)
````

Si el usuario escribe `4`, la interacción esperada es:

```text
Cantidad de cuadernos: 4
Cantidad comprada: 4
Total a pagar: 14000
```

Observe que la **entrada** es el valor escrito por el usuario y la **salida**
corresponde a los mensajes producidos por `print`.
